# Senti4SD Sentiment Classifier
This notebook runs Senti4SD sentiment analysis on Stack Overflow posts and merges the output with user and post metadata.


```python
import os
from subprocess import run

# Convert important paths to absolute (Windows-safe)
base_dir = os.path.abspath(".")

input_dir = os.path.abspath("datasets/preprocessed_for_senti4sd")
output_dir = os.path.abspath("datasets/predictions")
senti4sd_jar = os.path.abspath("external/pySenti4SD/java/Senti4SD-fast.jar")
dsm_file = os.path.abspath("external/pySenti4SD/java/dsm.bin")
model_file = os.path.abspath("external/pySenti4SD/Senti4SD.model")

os.makedirs(output_dir, exist_ok=True)

```

```python
csv_files = [f for f in os.listdir(input_dir) if f.endswith(".csv")]
total_files = len(csv_files)

print(f"📁 Total files to process: {total_files}\n")
```

```python
for idx, fname in enumerate(csv_files, start=1):
    input_path = os.path.abspath(os.path.join(input_dir, fname))
    output_path = os.path.abspath(os.path.join(output_dir, f"senti4sd_{fname}"))

    print(f"🔄 [{idx}/{total_files}] Running Senti4SD on: {fname}")

    cmd = [
        "java", "-Xmx16g", "-jar", senti4sd_jar,
        "-i", input_path,
        "-F", "A",
        "-W", dsm_file,
        "-oc", output_path,
        "-vd", "600",
        "-m", model_file
    ]

    print("📤 Command:", " ".join(cmd))  # Debug print

    result = run(cmd, capture_output=True, text=True)

    if result.returncode == 0:
        print(f"✅ Done: {fname} ({idx}/{total_files})\n")
    else:
        print(f"❌ Error with {fname}:\n{result.stderr}\n")
```
